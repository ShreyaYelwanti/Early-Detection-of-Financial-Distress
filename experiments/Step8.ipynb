```python
import pandas as pd

# Part 1: which features matter at each lead time
imp = pd.read_csv("results/step5_bins_importance.csv")
f = imp[(imp.level=="feature") & (imp.subset=="pos")]
t = f.pivot(index="name", columns="k_years_before", values="share")
print(t.assign(mx=t.max(axis=1)).sort_values("mx", ascending=False).head(12).round(3))

g = imp[(imp.level=="group") & (imp.subset=="pos")]
print(g[["k_years_before","name","share","share_lo","share_hi"]].round(3).to_string(index=False))

# Part 2: confidence intervals for the Step 4 model comparison
m = pd.read_csv("results/step4_metrics.csv")
m = m[(m.split=="test") & (m.features.isin(["full","-"]))]
print(m[["model","horizon","PR_AUC","PR_AUC_lo","PR_AUC_hi",
         "ROC_AUC","ROC_AUC_lo","ROC_AUC_hi"]].round(3).to_string(index=False))
print(pd.read_csv("results/step4_paired_vs_altman.csv")
      .query("features=='full'").round(3).to_string(index=False))
```
